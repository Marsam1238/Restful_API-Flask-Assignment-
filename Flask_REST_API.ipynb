# Theoretical Questions

### Q1. What is a RESTful API?

A RESTful API (Representational State Transfer) is a way of designing web services so that applications can communicate over the internet in a simple and lightweight manner. It uses standard HTTP methods like GET, POST, PUT, and DELETE to perform operations on resources, which are often represented in JSON. REST APIs are stateless, scalable, and easy to use, making them very popular for building modern web and mobile applications.

### Q2. Explain the concept of API specification

An API specification is like a rulebook that explains how an API should work. It defines the available endpoints, what kind of data you can send, what format the response will be in, and the possible status codes. With an API specification, developers know exactly how to interact with the API without guessing, which reduces confusion and ensures consistency.

### Q3. What is Flask, and why is it popular for building APIs?

Flask is a lightweight Python web framework that is often used to build APIs and small web applications. It is popular because it is simple to set up, easy to learn, and very flexible. Developers like Flask because it doesn’t force a strict structure—you can add only the parts you need. This makes it a great choice for small to medium projects or when you need a quick prototype.

### Q4. What is routing in Flask?

Routing in Flask means deciding what should happen when a user visits a certain URL. Each route is tied to a function that handles the request and sends back a response. For example, visiting `/home` might trigger a function that shows the homepage. Routing helps connect URLs to Python code so your app can respond to different requests.

### Q5. How do you create a simple Flask application?

To create a simple Flask app:
1. Install Flask with `pip install flask`.
2. Import Flask and create an app instance: `app = Flask(__name__)`.
3. Define routes using the `@app.route()` decorator.
4. Write functions to handle each route.
5. Run the server using `app.run()`. This will start a local web server where you can test your app.

### Q6. What are HTTP methods used in RESTful APIs?

The main HTTP methods used in RESTful APIs are:
- **GET**: Retrieve data.
- **POST**: Create new data.
- **PUT**: Update existing data.
- **DELETE**: Remove data.
- **PATCH**: Partially update data.

These methods help define what kind of operation you are performing on a resource.

### Q7. What is the purpose of the @app.route() decorator in Flask?

The `@app.route()` decorator tells Flask which URL should trigger a specific function. For example, `@app.route('/about')` means that when someone visits `/about`, Flask will run the function attached to that route. This makes it easy to link URLs to functionality in your app.

### Q8. What is the difference between GET and POST HTTP methods?

The **GET** method is used to fetch data from the server without changing anything. It is often cached and visible in the URL. The **POST** method is used to send data to the server, usually to create or update something. POST requests are not cached and the data is hidden in the request body.

### Q9. How do you handle errors in Flask APIs?

In Flask, you can handle errors using the `@app.errorhandler()` decorator or by raising exceptions. For example, if a page is not found, you can return a 404 error with a custom message. Common status codes include 400 (bad request), 404 (not found), and 500 (server error). Custom error messages can also be returned in JSON format to make APIs user-friendly.

### Q10. How do you connect Flask to a SQL database?

Flask can connect to a SQL database using extensions like Flask-SQLAlchemy. You install it with `pip install flask-sqlalchemy`, set up the database URI in your app configuration, and then create models that represent tables. With SQLAlchemy, you can run queries and manage your database using Python code instead of writing raw SQL.

### Q11. What is the role of Flask-SQLAlchemy?

Flask-SQLAlchemy is an extension that helps connect Flask apps to databases easily. It provides an ORM (Object Relational Mapper), which means you can work with Python classes and objects instead of writing raw SQL queries. This makes database operations simpler and more readable.

### Q12. What are Flask blueprints and how are they useful?

Blueprints in Flask are a way to organize your application into smaller, reusable parts. Each blueprint can have its own routes, templates, and static files. This is especially useful for larger apps, because it keeps the code organized and easier to maintain.

### Q13. What is the purpose of Flask's request object?

The `request` object in Flask holds all the information about the incoming HTTP request. You can use it to access form data, JSON payloads, query parameters, and headers. It is essential for handling any data that the client sends to your API.

### Q14. How do you create a RESTful API endpoint using Flask?

To create a RESTful API endpoint in Flask, you define a route with `@app.route()`, specify the allowed HTTP methods (GET, POST, etc.), and write a function that processes the request and returns a response, usually in JSON format.

### Q15. What is the purpose of Flask's jsonify() function?

The `jsonify()` function in Flask converts Python dictionaries or lists into JSON format and automatically sets the correct headers. This makes it easy to return JSON responses from your API without extra work.

### Q16. Explain Flask’s url_for() function

The `url_for()` function in Flask generates URLs dynamically based on the name of a route or function. Instead of hardcoding links, you can use `url_for()` to make your app more flexible and easier to maintain. For example, `url_for('home')` will give the URL of the `home` function’s route.

### Q17. How does Flask handle static files (CSS, JavaScript, etc.)?

Flask serves static files like CSS, JavaScript, and images from a folder named `static`. By default, these files can be accessed using `/static/filename`. You can link them in your HTML templates, and Flask will take care of serving them.

### Q21. How would you secure a Flask API?

To secure a Flask API, you can use techniques such as authentication (with tokens like JWT or OAuth), input validation, CORS handling, rate limiting, and encrypting sensitive data. Flask has extensions like Flask-JWT and Flask-OAuth that make adding these features easier.

### Q22. What is the significance of the Flask-RESTful extension?

Flask-RESTful is an extension that simplifies building REST APIs with Flask. It provides useful tools for defining resources, handling HTTP methods, parsing requests, and formatting responses. It reduces boilerplate code and helps developers create clean, structured APIs faster.

### Q23. What is the role of Flask’s session object?

The `session` object in Flask stores data across requests, usually for user sessions. It is often used to keep track of whether a user is logged in or to remember user preferences. The data is stored on the client side in cookies but is securely signed so it cannot be easily tampered with.

# Practical Questions

### Q1. How do you create a basic Flask library application?

```
from flask import Flask
app = Flask(__name__)

@app.route('/')
def home():
    return 'Hello, Flask!'

if __name__ == '__main__':
    app.run(debug=True)
```

### Q2. How do you serve static files like images or CSS in Flask framework?

```
from flask import Flask, send_from_directory
app = Flask(__name__)

@app.route('/static/<filename>')
def send_static(filename):
    return send_from_directory('static', filename)

if __name__ == '__main__':
    app.run(debug=True)
```

### Q3. How do you define different routes with different HyperText Transfer Protocol methods in Flask framework?

```
from flask import Flask
app = Flask(__name__)

@app.route('/greet', methods=['GET'])
def greet():
    return "Hello World!"

if __name__ == '__main__':
    app.run(debug=True)
```

### Q4. How do you render HTML templates in Flask framework?

```
from flask import Flask, render_template
app = Flask(__name__)

@app.route('/')
def index():
    return render_template('index.html')

if __name__ == '__main__':
    app.run(debug=True)
```

### Q5. How can you generate URLs for routes in Flask library using url_for?

```
from flask import Flask, url_for
app = Flask(__name__)

@app.route('/home')
def home():
    return 'This is the home page.'

@app.route('/')
def index():
    return f'<a href="{url_for("home")}">Go to Home</a>'

if __name__ == '__main__':
    app.run(debug=True)
```

### Q6. How do you handle forms in Flask framework?

```
from flask import Flask, request, render_template
app = Flask(__name__)

@app.route('/submit', methods=['GET', 'POST'])
def submit():
    if request.method == 'POST':
        name = request.form['name']
        return f"Hello {name}!"
    return render_template('form.html')

if __name__ == '__main__':
    app.run(debug=True)
```

### Q7. How can you validate form information in Flask library?

```
from flask import Flask, request, flash, redirect, url_for
app = Flask(__name__)
app.secret_key = 'secret'

@app.route('/login', methods=['POST'])
def login():
    username = request.form['username']
    if not username:
        flash('Username is required!')
        return redirect(url_for('login'))
    return 'Login successful'

if __name__ == '__main__':
    app.run(debug=True)
```

### Q8. How do you manage sessions in Flask library?

```
from flask import Flask, session
app = Flask(__name__)
app.secret_key = 'secret'

@app.route('/set_user')
def set_user():
    session['user'] = 'John'
    return 'User set'

@app.route('/get_user')
def get_user():
    return f'Logged in as {session.get("user", "Guest")}'

if __name__ == '__main__':
    app.run(debug=True)
```

### Q9. How do you redirect to a different route in Flask framework?

```
from flask import Flask, redirect, url_for
app = Flask(__name__)

@app.route('/redirect_home')
def redirect_home():
    return redirect(url_for('home'))

@app.route('/home')
def home():
    return 'This is the home page.'

if __name__ == '__main__':
    app.run(debug=True)
```

### Q10. How do you handle errors in Flask library (e.g., 404)?

```
from flask import Flask
app = Flask(__name__)

@app.errorhandler(404)
def page_not_found(e):
    return 'Page not found', 404

@app.route('/')
def home():
    return 'Welcome to the home page.'

if __name__ == '__main__':
    app.run(debug=True)
```

### Q11. How do you structure a Flask framework app using Blueprints?

```
from flask import Flask, Blueprint
app = Flask(__name__)

bp = Blueprint('home', __name__)

@bp.route('/')
def index():
    return 'Home Page'

app.register_blueprint(bp)

if __name__ == '__main__':
    app.run(debug=True)
```

### Q12. How do you define a custom Jinja filter in Flask library?

```
from flask import Flask
app = Flask(__name__)

@app.template_filter('reverse')
def reverse_filter(s):
    return s[::-1]

@app.route('/')
def home():
    return render_template('home.html', name='Flask')

if __name__ == '__main__':
    app.run(debug=True)
```

### 21. How do you handle file uploads in Flask framework?

```
from flask import Flask, request
app = Flask(__name__)

@app.route('/upload', methods=['GET', 'POST'])
def upload_file():
    if request.method == 'POST':
        file = request.files['file']
        file.save(f'./uploads/{file.filename}')
        return 'File uploaded successfully'
    return '''
        <form method="POST" enctype="multipart/form-data">
            <input type="file" name="file">
            <input type="submit">
        </form>
    '''

if __name__ == '__main__':
    app.run(debug=True)
```

### Q23. How do you customize error pages in Flask framework?

```
from flask import Flask, render_template
app = Flask(__name__)

@app.route('/')
def home():
    return 'Home Page'

@app.errorhandler(404)
def page_not_found(e):
    return render_template('404.html'), 404

if __name__ == '__main__':
    app.run(debug=True)
```
